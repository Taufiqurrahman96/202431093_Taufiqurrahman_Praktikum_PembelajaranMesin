```python
import os
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. muat data: pakai CSV kalau ada, kalau tidak pakai dataset bawaan scikit-learn
if os.path.exists('iris-flower.csv'):
    df = pd.read_csv('iris-flower.csv', sep=",")
    print("Data dimuat dari iris-flower.csv")
else:
    df = load_iris(as_frame=True).frame
    print("CSV tidak ditemukan, memakai dataset Iris bawaan scikit-learn")

print("Kolom:", df.columns.tolist())

# 2. cari kolom kelas dan kolom yang dibuang (id + kelas)
kol_kelas = next(c for c in df.columns if c.lower() in ('species', 'class', 'variety', 'target'))
buang = [c for c in df.columns if c.lower() in ('id', 'species', 'class', 'variety', 'target')]
print("Kolom kelas:", kol_kelas, "| dibuang dari fitur:", buang)

x = df.drop(columns=buang)
y = df[kol_kelas]

# 3. bagi data: 80% latih, 20% uji (proporsi spesies dijaga sama)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=0, stratify=y
)
print(x_train.shape, x_test.shape)

# 4. latih model
classifier = DecisionTreeClassifier(random_state=0)
classifier.fit(x_train, y_train)

# 5. prediksi dan evaluasi
y_pred = classifier.predict(x_test)
print("Akurasi:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import tree
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. muat data
if os.path.exists('iris-flower.csv'):
    df = pd.read_csv('iris-flower.csv', sep=",")
else:
    df = load_iris(as_frame=True).frame

# 2. fitur dan target (Id dan Species dibuang dari fitur)
buang = [c for c in df.columns if c.lower() in ('id', 'species', 'class', 'variety', 'target')]
kol_kelas = next(c for c in df.columns if c.lower() in ('species', 'class', 'variety', 'target'))

x = df.drop(columns=buang)
y = df[kol_kelas]

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=0, stratify=y
)

# 3. latih model
classifier = DecisionTreeClassifier(random_state=0)
classifier.fit(x_train, y_train)

# 4. gambar pohon keputusan
plt.subplots(figsize=(15, 10))
tree.plot_tree(
    classifier,
    feature_names=x.columns.tolist(),
    class_names=[str(k) for k in classifier.classes_],
    filled=True,
    fontsize=10
)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier(criterion="entropy", max_depth=5, random_state=42)
classifier.fit(x_train, y_train)
```

```python
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize=10) 
plt.show()
```

```python
depth = classifier.get_depth()
depth
```

```python
# Evaluasi

from sklearn.metrics import accuracy_score

y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```
